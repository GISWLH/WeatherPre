# WeatherPre · Earth2Studio models on a GPU (Colab)
Runs NVIDIA Earth2Studio prognostic models (AIFS v2, FourCastNet 3, Aurora 1.5, Atlas, U-CAST, GraphCast, FuXi-S2S, ...) from real initial conditions (ARCO ERA5 for past dates, GFS / IFS for the latest cycle), then scores them with WeatherBench-X next to the hosted models.

The WeatherPre side of this route is tested on CPU with Earth2Studio's `Persistence` model; the large models need a GPU runtime (*Runtime → Change runtime type → GPU*). Check each model's licence (several weights are non-commercial).

```
!pip -q install 'weatherpre[gpu] @ git+https://github.com/GISWLH/WeatherPre' 'earth2studio[fcn3]'   # swap extra: [aifs2] [aurora] [atlas] [ucast] [fuxi-s2s] ...
```

```
import weatherpre as wp
wp.models('weather').query("status == 'gpu'")
```

```
# weather scale: FourCastNet 3 from ERA5 initial conditions, compared with hosted GraphCast / Pangu / IFS
cmp = wp.compare(['fcn3', 'graphcast', 'pangu', 'ifs-hres'], 'z500,t2m', '7d', init='2020-10-03')
cmp.ranking()
```

```
cmp.plot('fcn3_scores.png'); cmp.plot_maps('fcn3_maps.png', 'z500', 120)
```

```
# any Earth2Studio prognostic class works directly
ds = wp.forecast('e2s:AIFS2', 't2m', '15d', init='latest')
```

```
# S2S scale: weekly means of a long rollout, compared with ECMWF extended range and baselines
s2s = wp.compare(['ucast', 'ifs-ext', 'climatology', 'persistence'], 't2m', '6w', init='2020-10-01')
s2s.table('acc')
```
